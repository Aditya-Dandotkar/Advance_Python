```python
from transformers import pipeline
import torch
from PIL import Image
import requests
```

```python
classifier = pipeline("image-classification",
                      model="google/vit-base-patch16-224"
                      )
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
image=Image.open("/content/drive/MyDrive/DS&AI traning/gettyimages-586890581.avif")
```

```python
result = classifier(image)
```

```python
for predictions in result[:5]:
  print(f"Label:{predictions['label']}")
  print(f"Confidence:{predictions['label']}")
  print(f"Confidence:{predictions['score']:.4f}")
  print("-"*40)
```
