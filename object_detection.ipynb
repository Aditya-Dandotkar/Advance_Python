```python
from transformers import pipeline
import torch
from PIL import Image
import requests
```

```python
classifier = pipeline("object-detection",
                      model="facebook/detr-resnet-50"
                      )
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
image=Image.open("/content/drive/MyDrive/BMW.jpeg")
```

```python
result=classifier(image)
```

```python
for predictions in result[:5]:
  print(f"Label:{predictions['label']}")
  print(f"Confidence:{predictions['label']}")
  print(f"Confidence:{predictions['score']:.4f}")
  print("-"*40)
```
